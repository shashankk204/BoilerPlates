```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain.schema.output_parser import StrOutputParser
from langchain_core.prompts import ChatPromptTemplate
from langchain.schema.runnable import RunnableLambda 
from dotenv import load_dotenv


import os
load_dotenv()

GOOGLE_API_KEY = os.getenv('GOOGLE_API_KEY')
llm=ChatGoogleGenerativeAI(model="gemini-2.0-flash",api_key=GOOGLE_API_KEY)

```

```python
prompt_template=ChatPromptTemplate(
    messages=[
        ("system","You are a good assistance"),
        ("human", "what is the current time in {location} according to India Standard Time?")
    ]
)

chain = prompt_template|llm|StrOutputParser()
```

```python
response=chain.invoke({"location":"Mississippi"})
```

# now using Agnets

```python
from langchain import hub
from langchain.agents import create_react_agent,AgentExecutor
```

```python
prompt_template=hub.pull("hwchase17/react")
```

```python
import datetime
from langchain.agents import tool

@tool
def get_system_time(format: str = "%Y-%m-%d %H:%M:%S"):
    """ Returns the current date and time of India in the specified format """

    current_time = datetime.datetime.now()
    formatted_time = current_time.strftime(format)
    return formatted_time



tools=[get_system_time]
```

```python
agent=create_react_agent(llm,tools,prompt_template)
Agent_Executer=AgentExecutor(agent=agent,tools=tools,verbose=True)

```

```python
Agent_Executer.invoke({"input":"what is the current time in Mississippi so if you have the current time in india you can easyly get the time at mississppi as you know the time diffrence ?"})
```
